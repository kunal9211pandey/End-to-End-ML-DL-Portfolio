```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("Salary_Experience.csv")
```

```python
df.head()
```

```python
x = df.iloc[: , 0:1].values
```

```python
y = df.iloc[ : , -1].values
```

```python
y
```

```python
plt.scatter(x,y)
plt.xlabel("years of Experiance")
plt.ylabel("Salary")
plt.title("Salary vs year of Experiance")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
```

```python
pf = PolynomialFeatures(degree = 5)
```

```python
lr = LinearRegression()
```

```python
x1 = pf.fit_transform(x)
```

```python
lr.fit(x1,y)
```

```python
lr.score(x1 ,y)
```

```python
plt.scatter(x,y, color = 'blue')
plt.plot(x , lr.predict(x1) , color = 'red')
plt.show()
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
import joblib
joblib.dump(lr , "Salary_experiance_predication.pkl")
```

```python
load = joblib.load("Salary_experiance_predication.pkl")
```
