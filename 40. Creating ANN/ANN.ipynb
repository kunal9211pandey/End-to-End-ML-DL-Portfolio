```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(iris.data, iris.target , test_size =0.2)
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
```

```python
input_layer = Dense(4, input_shape=(4,))
```

```python
model.add(input_layer)
```

```python
hidden_layer = Dense(4 , activation = 'relu')
```

```python
model.add(hidden_layer)
```

```python
output_layer = Dense(3 , activation='softmax')
```

```python
model.add(output_layer)
```

```python
# complie the model means load the model inot memory , model not run byut abilabel in memeory
model.compile(loss = 'sparse_categorical_crossentropy' , optimizer = 'adam' , metrics = ['accuracy'])
```

```python
# train the model
model.fit(x_train, y_train , epochs = 300 , batch_size=4)
```

```python
# Evaulate the model
val_loss , val_acc = model.evaluate(x_test, y_test)
```
