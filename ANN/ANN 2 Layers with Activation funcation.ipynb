```python
def relu(x):
    return (max(0 ,x))
```

```python
def sigmoid(x):
    if x <= 0:
        return 0
    else:
        return 1
        
```

```python
import numpy as np
```

```python
data = np.array([2,3])
```

```python
x0 = data[0]
x1 = data[1]
```

```python
input_layers = np.array([x0 , x1])
print("Input Layers = ", input_layers)
```

```python
# create  weights sum(xi * wi)
weight = { 'x2' : np.array([1,1]),
           'x3' : np.array([-1,1]),
           'x4' :np.array([-2,1]),
           'x5' :np.array([1,0]),
           'x6' : np.array([1, -1])
         }
```

```python
x2 = (input_layers * weight['x2']).sum()
x3 = (input_layers * weight['x3']).sum()
```

```python
x2 = relu(x2)
x3 = relu(x3)
```

```python
hidden_layer1 = np.array([x2 , x3])
print("Hidden Layers 1 = ", hidden_layer1)
```

```python
x4 = (hidden_layer1 * weight['x4']).sum()
x5 = (hidden_layer1 * weight['x5']).sum()
```

```python
x4 = relu(x4)
x5 = relu(x5)
```

```python
hidden_layers2 = np.array([x4 , x5])
print("Hidden layers 2 = ", hidden_layers2)
```

```python
x6 = (hidden_layers2 * weight['x6']).sum()
```

```python
x6 = sigmoid(x6)
```

```python
output_layers = np.array([x6])
```

```python
print("the final output layers is ", output_layers)
```
