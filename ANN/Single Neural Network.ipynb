```python
import numpy as np
```

# create Input Layers 

```python
data = np.array([2,3])
```

```python
data
```

```python
x0 = data[0]
x1 = data[1]
```

```python
input_layers = np.array([x0 , x1])
print("Input Layers = ", input_layers)
```

# Create Hidden Layers 

```python
# create hidden layers
weight = { 'x2' : np.array([1,1]),
           'x3' : np.array([-1,1]),
           'x4' :np.array([2,-1])}
```

```python
weight
```

```python
x2 = (input_layers * weight['x2']).sum()
x3 = (input_layers * weight['x3']).sum()
```

```python
x2
```

```python
x3
```

```python
hidden_layers = np.array([x2,x3])
print("the hidden layers is ",hidden_layers)
```

# create Output Layers 

```python
# create Output layers
x4 = (hidden_layers * weight['x4']).sum()
```

```python
x4
```

```python
output_layers = np.array([x4])
```

```python
output_layers
```
