```python
a = tf.constant(10)
b = tf.constant(20)
c = a + b

print(c)
print(c.numpy())
```

```python
add = tf.add(a, b)
sub = tf.subtract(a, b)
mul = tf.multiply(a, b)
div = tf.divide(a, b)
power = tf.pow(a, b)
# exp = tf.exp(a)
sqr = tf.sqrt(tf.cast(a, tf.float32))  # sqrt needs float

print("Add:", add.numpy())
print("Subtract:", sub.numpy())
print("Multiply:", mul.numpy())
print("Divide:", div.numpy())
print("Power:", power.numpy())
# print("Exp:", exp.numpy())
print("Sqrt:", sqr.numpy())
```
