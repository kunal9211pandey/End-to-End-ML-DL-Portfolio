# How to remove the columns

```python
data1 = data.copy()
```

```python
data1.drop(['MSRP','Invoice'], axis = 1, inplace=True)
```

```python
data1.head(5)
```

# check duplicate and null values

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(keep="first" ,inplace=True)
```

```python
data.duplicated().sum()
```

```python
data['MSRP'] = data['MSRP'].replace('[\$,]', '' ,regex=True).str.strip()
data['Invoice'] = data['Invoice'].replace('[\$,]', '',regex=True).str.strip()
```

```python
data.head()
```

```python
data['Invoice'] = pd.to_numeric(data['Invoice'])
data['MSRP'] = pd.to_numeric(data['MSRP'])
```

# convert doller to Indian Money

```python
usd_to_inr = 90.70
data['MSRP_INR'] = data['MSRP'] * usd_to_inr
data['Invoice_INR'] = data['Invoice'] * usd_to_inr
```

```python
data.head()
```

```python
data['MSRP_INR'] = data['MSRP_INR'].round(0)
data['Invoice_INR'] = data['Invoice_INR'].round(0)
```

```python
data[['MSRP', 'MSRP_INR', 'Invoice', 'Invoice_INR']].head()
```

```python
data['MSRP_INR'] = data['MSRP_INR'].astype(int)
data['Invoice_INR'] = data['Invoice_INR'].astype(int)
```

```python
data[['MSRP', 'MSRP_INR', 'Invoice', 'Invoice_INR']].head()
```

```python
data.drop(['MSRP' , 'Invoice'] , axis = 1 , inplace=True)
```

```python
conditions = [
    data['MSRP_INR'] < 2000000,
    (data['MSRP_INR'] >= 2000000) & (data['MSRP_INR'] < 4000000),
    (data['MSRP_INR'] >= 4000000) & (data['MSRP_INR'] < 8000000),
    data['MSRP_INR'] >= 8000000
]

labels = ['Budget', 'Mid_range', 'Premium', 'Luxury']

data['Price_Segment'] = np.select(conditions, labels)
```

```python
data['Price_Segment'].value_counts()
```

```python
data['Price_Segment'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(5,5)
)

plt.title('Market Share by Price Segment')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8,5))

avg_price = data.groupby('Type')['MSRP_INR'].mean().sort_values()

sns.barplot(x=avg_price.index, y=avg_price.values)

plt.title('Average Price by Car Type')
plt.xlabel('Car Type')
plt.ylabel('Average MSRP (INR)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Type', hue='Price_Segment', data=data)

plt.title('Price Segment Distribution by Car Type')
plt.show()
```

```python
#Origin (Asia, USA, Europe) ke hisaab se average price me kya difference hai?
data.groupby('Origin')['MSRP_INR'].mean().round(0).astype(int)
```

```python
origin_price_lakh = (data.groupby('Origin')['MSRP_INR'].mean() / 100000).round(2)

plt.figure(figsize=(6,4))
sns.barplot(x=origin_price_lakh.index, y=origin_price_lakh.values)

plt.title('Average Car Price by Origin (in Lakhs)')
plt.xlabel('Origin')
plt.ylabel('Average Price (Lakhs)')
plt.show()
```

```python
# DriveTrain (Front, Rear, All) ka price par kya effect hai?
drivetrain_price = data.groupby('DriveTrain')['MSRP_INR'].mean().round(0).astype(int)
drivetrain_price
```

```python
drivetrain_price_lakh = (data.groupby('DriveTrain')['MSRP_INR'].mean() / 100000).round(2)

plt.figure(figsize=(6,4))
sns.barplot(x=drivetrain_price_lakh.index, y=drivetrain_price_lakh.values)

plt.title('Average Price by DriveTrain (in Lakhs)')
plt.xlabel('DriveTrain')
plt.ylabel('Average Price (Lakhs)')
plt.show()
```

```python
drive_count = data['DriveTrain'].value_counts()

plt.figure(figsize=(5,5))
drive_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Car Distribution by DriveTrain')
plt.ylabel('')
plt.show()
```

```python
drive_price = data.groupby('DriveTrain')['MSRP_INR'].sum()

plt.figure(figsize=(5,5))
drive_price.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Total Price Share by DriveTrain')
plt.ylabel('')
plt.show()
```

```python
# EngineSize aur Price ke beech kya relationship hai?
plt.figure(figsize=(7,5))
sns.scatterplot(x=data['EngineSize'], y=data['MSRP_INR']/100000)

plt.title('Engine Size vs Price (in Lakhs)')
plt.xlabel('Engine Size')
plt.ylabel('Price (Lakhs)')
plt.show()
```

```python
data['Engine_Bin'] = pd.cut(data['EngineSize'], bins=[1,2,3,4,5,6,9])

avg_price = data.groupby('Engine_Bin')['MSRP_INR'].mean()/100000

plt.figure(figsize=(7,5))
sns.barplot(x=avg_price.index.astype(str), y=avg_price.values)

plt.title('Average Price by Engine Size Range')
plt.xlabel('Engine Size Range')
plt.ylabel('Average Price (Lakhs)')
plt.show()
```
