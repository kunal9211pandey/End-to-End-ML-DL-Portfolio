```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("cricket.csv")
```

```python
df.head(10)
```

```python
# convert data into numeric formate
def outlook_to_numeric(x):
    if x == 'Rainy':
        return 0
    elif x == 'Overcast':
        return 1
    elif x == 'Sunny':
        return 2
```

```python
def temp_to_numeric(x):
    if x == 'Hot':
        return 0
    elif x == 'Mild':
        return 1
    elif x == 'Cool':
        return 2
```

```python
# convert data into numeric formate
def HUMIDITY_to_numeric(x):
    if x == 'High':
        return 0
    elif x == 'Normal':
        return 1
```

```python
# convert data into numeric formate
def WINDY_to_numeric(x):
    if x == False:
        return 0
    elif x == True:
        return 1
```

```python
# convert data into numeric formate
def PLAY_CRICKET_to_numeric(x):
    if x == 'Yes':
        return 0
    elif x == 'No':
        return 1
```

```python
df.columns
```

```python
df['OUTLOOK'] = df['OUTLOOK'].apply(outlook_to_numeric)
df['TEMPERATURE'] = df['TEMPERATURE'].apply(temp_to_numeric)
df['HUMIDITY'] = df['HUMIDITY'].apply(HUMIDITY_to_numeric)
df['WINDY'] = df['WINDY'].apply(WINDY_to_numeric)
df['PLAY CRICKET'] = df['PLAY CRICKET'].apply(PLAY_CRICKET_to_numeric)
```

```python
df.head()
```

```python
x = df[['OUTLOOK', 'TEMPERATURE', 'HUMIDITY', 'WINDY']]
```

```python
x
```

```python
y = df['PLAY CRICKET']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x, y, test_size = 0.3 , random_state = 10)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(x_train , y_train)
```

```python
y_pred = gnb.predict(x_test)
```

```python
y_pred
```

```python
from sklearn import metrics
```

```python
score = metrics.accuracy_score(y_test, y_pred)
```
