```python
import pandas as pd
```

```python
df = pd.read_csv("smsspamcollection/SMSSpamCollection", sep='\t', header=None, names=['Status', 'Message'])

df.head()
```

```python
len(df)
```

```python
len(df[df.Status == 'spam'])
```

```python
len(df[df.Status == 'ham'])
```

```python
# convert spam as 0 and ham is 1
df.loc[df["Status"] == 'spam' , "status"] = 0
df.loc[df["Status"] == 'ham' , "status"] = 1
```

```python
df.head()
```

```python
df['Status'] = df['Status'].map({'spam': 0, 'ham': 1})
```

```python
df.head()
```

```python
df.drop('status', axis=1, inplace=True)
```

```python
df.head()
```

```python
x = df["Message"]
y = df["Status"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y ,test_size =0.3 )
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
cv = CountVectorizer()
```

```python
# count the frequecny of the word in dataset 
x_traincv = cv.fit_transform(x_train)
x_testcv = cv.transform(x_test)
```

```python
# befroe train the model take y as int
y_train = y_train.astype('int')
y_test = y_test.astype('int')
```

```python
mnb = MultinomialNB()
```

```python
mnb.fit(x_traincv , y_train)
```

```python
mnb.score(x_testcv , y_test)
```
