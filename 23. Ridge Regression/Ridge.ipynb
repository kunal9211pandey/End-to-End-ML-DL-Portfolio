```python
import pandas as pd
```

```python
df = pd.read_csv("train.csv")
```

```python
df
```

```python
df1 = pd.read_csv("test.csv")
```

```python
df1
```

```python
x_train = df.iloc[: , 0:1].values
y_train = df.iloc[ : , 1].values
```

```python
x_test = df1.iloc[: , 0:1].values
y_test = df1.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt

# Train data
plt.scatter(x_train, y_train, color='blue', label='Train Data')

# Test data
plt.scatter(x_test, y_test, color='red', label='Test Data')
plt.plot(x_train, y_train, color='blue')
plt.plot(x_test, y_test, color='red')

# Labels
plt.xlabel("Weight")
plt.ylabel("Height")

# Legend show karega kaunsa train hai kaunsa test
plt.legend()

# Graph show
plt.show()
```

```python
import numpy as np

# Scatter points
plt.scatter(x_train, y_train, color='blue', label='Train Data')
plt.scatter(x_test, y_test, color='red', label='Test Data')

# -------- Train line --------
slope_train, intercept_train = np.polyfit(x_train.flatten(), y_train, 1)

x_train_line = np.linspace(x_train.min(), x_train.max(), 100)
y_train_line = slope_train * x_train_line + intercept_train

plt.plot(x_train_line, y_train_line, color='blue', linestyle='--', label='Train Line')

# -------- Test line --------
slope_test, intercept_test = np.polyfit(x_test.flatten(), y_test, 1)

x_test_line = np.linspace(x_test.min(), x_test.max(), 100)
y_test_line = slope_test * x_test_line + intercept_test

plt.plot(x_test_line, y_test_line, color='red', linestyle='--', label='Test Line')

# Labels
plt.xlabel("Weight")
plt.ylabel("Height")
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train , y_train)
```

```python
from sklearn.metrics import r2_score
lr.score(x_train , y_train)
```

```python
plt.scatter(x_train , y_train , color = 'blue')
plt.plot(x_train ,lr.predict(x_train) ,color = 'red')
plt.show()
```

```python
rr = Ridge(alpha = 0.001)
rr.fit(x_test , y_test)
```

```python
rr.score(x_test, y_test)
```

```python
plt.scatter(x_test , y_test , color = 'red')
plt.plot(x_test ,rr.predict(x_test) ,color = 'green')
plt.show()
```
