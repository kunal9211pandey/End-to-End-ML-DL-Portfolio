```python
import pandas as pd
```

```python
df = pd.read_csv("boston_houses.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.rename({'Unnamed: 0' :'a'} , axis = 'columns' , inplace = True)
```

```python
df.head()
```

```python
df.drop(['a'] , axis = 1 , inplace = True)
```

```python
df.head()
```

```python
x = df.drop('MEDV',axis = 1)
y = df['MEDV'].values
```

```python
name = x.columns
```

```python
rng =range(len(name))
```

```python
rng
```

```python
from sklearn.linear_model import Lasso
ls = Lasso(alpha = 0.1)
```

```python
data = ls.fit(x,y)
```

```python
cf = data.coef_
```

```python
cf
```

```python
import matplotlib.pyplot as plt
plt.plot(rng ,cf)
plt.show()
```

```python
data.score(x , y)
```
