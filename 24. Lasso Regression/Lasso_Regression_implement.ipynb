```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df  = pd.read_csv("Advertising.csv")
```

```python
df.head()
```

```python
x = df.drop('sales' , axis = 1)
x
```

```python
y = df['sales'].values
```

```python
from sklearn.linear_model import Lasso
le = Lasso(alpha = 0.01)
```

```python
model = le.fit(x ,y)
```

```python
cf = model.coef_
```

```python
cf
```

```python
name = x.columns
```

```python
rng = range(len(x.columns))
```

```python
rng
```

```python
plt.plot(rng , cf)
plt.xticks(rng, name , rotation = 60)
plt.ylabel("cofficent")
plt.show()
```

```python
le.score(x,y)
```
