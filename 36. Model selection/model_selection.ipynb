```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
columns = ['age' ,'sex' , 'cp' ,'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'num']
```

```python
df = pd.read_csv("processed.cleveland.data" , sep =',', names = columns)
```

```python
df.head()
```

```python
# convert nums value into 0 or 1
df['num'] = df.num.map({0:0 , 1:1,2:1, 3:1, 4:1})
```

```python
df.head()
```

```python
x = df.iloc[: , :-1].values
y= df.iloc[: ,-1].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state = 0)
```

```python
# apply support vector machine
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train1 = scaler.fit_transform(x_train)
x_test1 = scaler.transform(x_test)
```

```python
from sklearn.svm import SVC
model = SVC(kernel = 'rbf')
model.fit(x_train1, y_train)
test_score = model.score(x_test1 , y_test)
train_score = model.score(x_train1 , y_train)
```

```python
test_score
```

```python
train_score
```

```python
# apply naive Bayes
from sklearn.naive_bayes import GaussianNB
model2 = GaussianNB()
model2.fit(x_train, y_train)
train_score = model2.score(x_train , y_train)
test_score = model2.score(x_test, y_test)
```

```python
train_score
```

```python
test_score
```

```python
# apply logistic regression
from sklearn.linear_model import LogisticRegression
model3 = LogisticRegression(solver = 'liblinear')
model3.fit(x_train, y_train)
model3.score(x_test, y_test)
```

```python
model3.score(x_train, y_train)
```

```python
# decision Tree
from sklearn.tree import DecisionTreeClassifier
model4 = DecisionTreeClassifier()
model4.fit(x_train, y_train)
model4.score(x_test, y_test)
```

```python
model4.score(x_train, y_train)
```

```python
# apply random forest 
from sklearn.ensemble import RandomForestClassifier
model5 = RandomForestClassifier()
model5.fit(x_train, y_train)
model5.score(x_test, y_test)
```

```python
model5.score(x_train, y_train)
```
