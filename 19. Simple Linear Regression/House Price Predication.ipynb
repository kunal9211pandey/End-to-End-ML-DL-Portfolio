```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("homeprices.csv")
```

```python
data
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
sns.scatterplot(data= data , x='area' , y='price')
plt.show()
```

```python
reg = LinearRegression()
```

```python
reg.fit(data[['area']] , data.price)
```

```python
reg.predict([[3300]])
```

```python
# find the cofficent this is slop m
reg.coef_
```

```python
# find intercept this is b
reg.intercept_
```

```python
y = 135.78767123 * 3300 + 180616.43835616432
y
```

```python
reg.predict([[5500]])
```

```python
from sklearn.metrics import r2_score
```

```python
y_original = data.price
y_predict = reg.predict(data[['area']])
R_score = r2_score(y_original , y_predict)
R_score
```

```python
sns.lmplot(data = data , x = 'area' , y='price')
plt.show()
```

```python
import joblib

# trained model save
joblib.dump(reg, 'house_price_linear_model.pkl')

print("Model saved successfully")
```
