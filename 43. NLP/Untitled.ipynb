```python
# stemming :- normlize the word into thier base or root called staming
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem('having')
```

```python
word = ['give' , 'giving', 'given' , 'gave']
for w in word:
    print(w," : ", pst.stem(w))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
```

```python
for w in word:
    print(w , " : ", lst.stem(w))
```

```python
# lemmatization , steming not give root word many time , but lemma group together diffenrt inflicted from of the same word called leema
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()
```

```python
wnl.lemmatize('corpora')
```

```python
for w in word:
    print(w , " : " ,wnl.lemmatize(w))
```

```python
for w in word:
    print(w , " : " ,wnl.lemmatize(w , pos = 'v'))
```

```python
# tf-idf (term frequency - inverse document frequency) value for each word
# tf how word occure in sentence , if several occureance of the same word in the same sentace, 
#so tf is increaed , mens word reapt many time i the sentace not foudn other senetend  so tf given more important

# idf : how common is word accross many sentence , if word is used many sentence then tf-idf decresed, some common word may repated in senetece so they
# give more importangt
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "machine learning is fun",
    "deep learning is powerful",
    "machine learning and deep learning"
]
```

```python
vectorizer = TfidfVectorizer()
```

```python
X = vectorizer.fit_transform(documents)
```

```python
print(X)
```

```python
print(vectorizer.get_feature_names_out())
print(X.toarray())
```
